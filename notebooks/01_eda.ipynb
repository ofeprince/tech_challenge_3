# 1. Análise de Valores Ausentes (Missing Values)

Antes de qualquer análise, precisamos entender a distribuição dos valores nulos para decidir estratégias de preenchimento no feature engineering.

```python
# Análise de valores nulos por coluna
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Total Nulos': missing,
    '% Nulos': missing_pct
})
missing_df = missing_df[missing_df['Total Nulos'] > 0]
print(missing_df.to_string())
```

```python
# Visualização dos valores nulos
fig, ax = plt.subplots(figsize=(10, 6))

# Gráfico de barras com % de nulos
missing_df['% Nulos'].plot(kind='barh', ax=ax, color='salmon')

ax.set_title('Percentual de Valores Nulos por Coluna', fontsize=13)
ax.set_xlabel('% de Nulos')
ax.axvline(x=50, color='red', linestyle='--', label='50%')
ax.legend()

plt.tight_layout()
plt.show()

print(f"\nObservação: As colunas de delay (AIR_SYSTEM_DELAY, SECURITY_DELAY, etc.) possuem ~{missing_df.loc['AIR_SYSTEM_DELAY', '% Nulos']:.0f}% de nulos.")
print("Isso ocorre pois só são preenchidas quando existe atraso. O restante dos nulos é de voos cancelados/desviados.")
```

# 2. Voos Cancelados e Desviados

Voos cancelados e desviados não possuem `ARRIVAL_DELAY`. Para o modelo de atraso, precisamos tratá-los separadamente.

```python
# Distribuição geral dos status dos voos
total = len(df)
cancelled = df['CANCELLED'].sum()
diverted = df['DIVERTED'].sum()
normal = total - cancelled - diverted

print("=" * 50)
print(f"Total de voos:       {total:>10,}")
print(f"Voos normais:        {normal:>10,}  ({normal/total*100:.2f}%)")
print(f"Voos cancelados:     {cancelled:>10,}  ({cancelled/total*100:.2f}%)")
print(f"Voos desviados:      {diverted:>10,}  ({diverted/total*100:.2f}%)")
print("=" * 50)

# Motivos de cancelamento
cancel_reasons = {
    'A': 'Companhia Aérea',
    'B': 'Clima',
    'C': 'NAS (Sistema Nacional)',
    'D': 'Segurança'
}

cancel_counts = df[df['CANCELLED'] == 1]['CANCELLATION_REASON'].value_counts()
cancel_counts.index = cancel_counts.index.map(cancel_reasons)

# Criação do gráfico
fig, ax = plt.subplots(figsize=(10, 5))

cancel_counts.plot(
    kind='barh',
    ax=ax,
    color=['#e74c3c', '#3498db', '#f39c12', '#9b59b6'],
    edgecolor='black'
)

ax.set_title('Motivos de Cancelamento', fontsize=13)
ax.set_xlabel('Quantidade de Voos')
ax.set_ylabel('')

# Adiciona valores nas barras
for p in ax.patches:
    ax.annotate(
        f'{p.get_width():,.0f}',
        (p.get_width(), p.get_y() + p.get_height()/2),
        ha='left',
        va='center',
        fontsize=9,
        xytext=(5, 0),
        textcoords='offset points'
    )

plt.tight_layout()
plt.show()
```

# 3. Variável Alvo: ARRIVAL_DELAY

A FAA (Federal Aviation Administration) define atraso como chegada **≥ 15 minutos** após o horário previsto. Vamos usar essa definição para criar nossa variável alvo binária `IS_DELAYED`.

```python
# Filtrar apenas voos não cancelados e não desviados com ARRIVAL_DELAY disponível
df_flights = df[
    (df['CANCELLED'] == 0) &
    (df['DIVERTED'] == 0) &
    (df['ARRIVAL_DELAY'].notna())
].copy()

# Criar variável alvo binária (critério FAA: >= 15 minutos)
df_flights['IS_DELAYED'] = (df_flights['ARRIVAL_DELAY'] >= 15).astype(int)

delayed = df_flights['IS_DELAYED'].sum()
on_time = len(df_flights) - delayed

print(f"Dataset para modelagem: {len(df_flights):,} voos")
print(f"  Pontual (< 15 min):   {on_time:>10,}  ({on_time/len(df_flights)*100:.2f}%)")
print(f"  Atrasado (>= 15 min): {delayed:>10,}  ({delayed/len(df_flights)*100:.2f}%)")

# Apenas 2 gráficos agora
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Balanceamento da classe alvo
axes[0].bar(
    ['Pontual', 'Atrasado'],
    [on_time, delayed],
    color=['#2ecc71', '#e74c3c'],
    edgecolor='black'
)

axes[0].set_title('Balanceamento da Variável Alvo (IS_DELAYED)', fontsize=12)
axes[0].set_ylabel('Quantidade de Voos')

for p in axes[0].patches:
    axes[0].annotate(
        f'{p.get_height()/len(df_flights)*100:.1f}%',
        (p.get_x() + p.get_width()/2., p.get_height()),
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

# Distribuição do ARRIVAL_DELAY (sem outliers extremos)
delay_data = df_flights['ARRIVAL_DELAY'].clip(-60, 180)

axes[1].hist(
    delay_data,
    bins=100,
    color='#3498db',
    edgecolor='black',
    alpha=0.7
)

axes[1].axvline(
    x=15,
    color='red',
    linestyle='--',
    linewidth=2,
    label='Limiar FAA (15 min)'
)

axes[1].axvline(
    x=0,
    color='orange',
    linestyle='--',
    linewidth=1.5,
    label='Zero'
)

axes[1].set_title(
    'Distribuição do ARRIVAL_DELAY (clip -60 a 180 min)',
    fontsize=12
)

axes[1].set_xlabel('Minutos de Atraso na Chegada')
axes[1].set_ylabel('Frequência')
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"\nEstatísticas do ARRIVAL_DELAY:")
print(df_flights['ARRIVAL_DELAY'].describe().round(2).to_string())
```

# 4. Análise Temporal: Atrasos por Mês, Dia da Semana e Hora

Padrões temporais são fortes preditores de atraso e serão features importantes no modelo.

```python
# Atraso médio e % de atrasos por mês
month_names = {1:'Jan', 2:'Fev', 3:'Mar', 4:'Abr', 5:'Mai', 6:'Jun',
               7:'Jul', 8:'Ago', 9:'Set', 10:'Out', 11:'Nov', 12:'Dez'}

by_month = df_flights.groupby('MONTH').agg(
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
by_month['MONTH_NAME'] = by_month['MONTH'].map(month_names)
by_month['pct_delayed'] *= 100

# Atraso por dia da semana
day_names = {1:'Seg', 2:'Ter', 3:'Qua', 4:'Qui', 5:'Sex', 6:'Sab', 7:'Dom'}
by_dow = df_flights.groupby('DAY_OF_WEEK').agg(
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
by_dow['DAY_NAME'] = by_dow['DAY_OF_WEEK'].map(day_names)
by_dow['pct_delayed'] *= 100

# Extrair hora da partida programada (HHMM -> HH)
df_flights['SCHED_HOUR'] = (df_flights['SCHEDULED_DEPARTURE'] // 100).astype(int)
by_hour = df_flights.groupby('SCHED_HOUR').agg(
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean'),
    count=('ARRIVAL_DELAY', 'count')
).reset_index()
by_hour['pct_delayed'] *= 100

fig, axes = plt.subplots(2, 3, figsize=(20, 10))

# Atraso médio por mês
axes[0, 0].bar(by_month['MONTH_NAME'], by_month['avg_delay'], color='#3498db', edgecolor='black')
axes[0, 0].set_title('Atraso Médio (min) por Mês', fontsize=12)
axes[0, 0].set_ylabel('Minutos de Atraso')
axes[0, 0].axhline(y=df_flights['ARRIVAL_DELAY'].mean(), color='red', linestyle='--', label='Média geral')
axes[0, 0].legend()

# % de atrasos por mês
axes[0, 1].bar(by_month['MONTH_NAME'], by_month['pct_delayed'], color='#e74c3c', edgecolor='black')
axes[0, 1].set_title('% de Voos Atrasados por Mês', fontsize=12)
axes[0, 1].set_ylabel('% Atrasados')
axes[0, 1].axhline(y=by_month['pct_delayed'].mean(), color='navy', linestyle='--', label='Média geral')
axes[0, 1].legend()

# Volume de voos por mês
vol_month = df_flights.groupby('MONTH').size().reset_index(name='count')
vol_month['MONTH_NAME'] = vol_month['MONTH'].map(month_names)
axes[0, 2].bar(vol_month['MONTH_NAME'], vol_month['count'], color='#95a5a6', edgecolor='black')
axes[0, 2].set_title('Volume de Voos por Mês', fontsize=12)
axes[0, 2].set_ylabel('Número de Voos')

# Atraso médio por dia da semana
axes[1, 0].bar(by_dow['DAY_NAME'], by_dow['avg_delay'], color='#9b59b6', edgecolor='black')
axes[1, 0].set_title('Atraso Médio (min) por Dia da Semana', fontsize=12)
axes[1, 0].set_ylabel('Minutos de Atraso')
axes[1, 0].axhline(y=df_flights['ARRIVAL_DELAY'].mean(), color='red', linestyle='--', label='Média geral')
axes[1, 0].legend()

# % de atrasos por dia da semana
axes[1, 1].bar(by_dow['DAY_NAME'], by_dow['pct_delayed'], color='#e67e22', edgecolor='black')
axes[1, 1].set_title('% de Voos Atrasados por Dia da Semana', fontsize=12)
axes[1, 1].set_ylabel('% Atrasados')

# Atraso médio por hora do dia
axes[1, 2].plot(by_hour['SCHED_HOUR'], by_hour['avg_delay'], marker='o', color='#2c3e50', linewidth=2)
axes[1, 2].fill_between(by_hour['SCHED_HOUR'], by_hour['avg_delay'], alpha=0.3, color='#3498db')
axes[1, 2].set_title('Atraso Médio por Hora do Dia (Partida Programada)', fontsize=12)
axes[1, 2].set_xlabel('Hora do Dia')
axes[1, 2].set_ylabel('Minutos de Atraso')
axes[1, 2].axhline(y=0, color='gray', linestyle='--')

plt.tight_layout()
plt.show()

print("Meses com mais atrasos:", by_month.nlargest(3, 'pct_delayed')[['MONTH_NAME','pct_delayed']].to_string(index=False))
print("\nDias com mais atrasos:", by_dow.nlargest(3, 'pct_delayed')[['DAY_NAME','pct_delayed']].to_string(index=False))
```

```python
# Merge com nomes das companhias aéreas
df_flights_airlines = df_flights.merge(df_airlines, left_on='AIRLINE', right_on='IATA_CODE', how='left')

by_airline = df_flights_airlines.groupby('AIRLINE_y').agg(
    total_flights=('ARRIVAL_DELAY', 'count'),
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean'),
    avg_dep_delay=('DEPARTURE_DELAY', 'mean')
).reset_index()
by_airline.columns = ['AIRLINE', 'TOTAL_FLIGHTS', 'AVG_ARRIVAL_DELAY', 'PCT_DELAYED', 'AVG_DEP_DELAY']
by_airline['PCT_DELAYED'] *= 100
by_airline = by_airline.sort_values('PCT_DELAYED', ascending=True)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# % de atrasos por companhia
colors = ['#e74c3c' if x > by_airline['PCT_DELAYED'].mean() else '#2ecc71' for x in by_airline['PCT_DELAYED']]
axes[0].barh(by_airline['AIRLINE'], by_airline['PCT_DELAYED'], color=colors, edgecolor='black')
axes[0].axvline(x=by_airline['PCT_DELAYED'].mean(), color='navy', linestyle='--', label='Média')
axes[0].set_title('% de Voos Atrasados por Companhia', fontsize=12)
axes[0].set_xlabel('% Atrasados')
axes[0].legend()

# Atraso médio de chegada por companhia
by_airline_sorted_delay = by_airline.sort_values('AVG_ARRIVAL_DELAY', ascending=True)
axes[1].barh(by_airline_sorted_delay['AIRLINE'], by_airline_sorted_delay['AVG_ARRIVAL_DELAY'],
             color=['#e74c3c' if x > 0 else '#2ecc71' for x in by_airline_sorted_delay['AVG_ARRIVAL_DELAY']],
             edgecolor='black')
axes[1].axvline(x=0, color='black', linewidth=1)
axes[1].set_title('Atraso Médio de Chegada (min) por Companhia', fontsize=12)
axes[1].set_xlabel('Minutos')

# Volume de voos por companhia
by_airline_vol = by_airline.sort_values('TOTAL_FLIGHTS', ascending=True)
axes[2].barh(by_airline_vol['AIRLINE'], by_airline_vol['TOTAL_FLIGHTS'] / 1000,
             color='#3498db', edgecolor='black')
axes[2].set_title('Volume de Voos por Companhia (milhares)', fontsize=12)
axes[2].set_xlabel('Milhares de Voos')

plt.tight_layout()
plt.show()

print("\nRanking de pontualidade (menor % atrasado → mais pontual):")
print(by_airline[['AIRLINE', 'TOTAL_FLIGHTS', 'AVG_ARRIVAL_DELAY', 'PCT_DELAYED']]
      .sort_values('PCT_DELAYED')
      .to_string(index=False))
```

# 6. Tipos de Atraso: Causas Raiz

O dataset contém 5 categorias de causa de atraso. Entender a composição ajuda a construir features mais informativas.

```python
# Voos com pelo menos uma causa de atraso registrada
delay_cols = ['AIR_SYSTEM_DELAY', 'SECURITY_DELAY', 'AIRLINE_DELAY', 'LATE_AIRCRAFT_DELAY', 'WEATHER_DELAY']
delay_labels = ['Sistema Aéreo', 'Segurança', 'Companhia', 'Aeronave Atrasada', 'Clima']

df_with_cause = df_flights[df_flights[delay_cols].notna().any(axis=1)].copy()
df_with_cause[delay_cols] = df_with_cause[delay_cols].fillna(0)

print(f"Voos com causa de atraso registrada: {len(df_with_cause):,} ({len(df_with_cause)/len(df_flights)*100:.1f}% do total)")

# Contribuição média de cada tipo
mean_delays = df_with_cause[delay_cols].mean()
total_mean = mean_delays.sum()

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Pizza: proporção de cada causa
axes[0].pie(mean_delays, labels=delay_labels, autopct='%1.1f%%', startangle=90,
            colors=['#3498db', '#9b59b6', '#e74c3c', '#f39c12', '#1abc9c'])
axes[0].set_title('Proporção de Cada Tipo de Atraso\n(sobre voos com causa registrada)', fontsize=12)

# Barras: minutos médios de cada causa
axes[1].bar(delay_labels, mean_delays.values, color=['#3498db', '#9b59b6', '#e74c3c', '#f39c12', '#1abc9c'],
            edgecolor='black')
axes[1].set_title('Contribuição Média (min) por Tipo de Atraso', fontsize=12)
axes[1].set_ylabel('Minutos Médios')
axes[1].tick_params(axis='x', rotation=20)

# % de voos afetados por cada tipo
pct_affected = (df_with_cause[delay_cols] > 0).mean() * 100
axes[2].bar(delay_labels, pct_affected.values, color=['#3498db', '#9b59b6', '#e74c3c', '#f39c12', '#1abc9c'],
            edgecolor='black')
axes[2].set_title('% de Voos Afetados por Cada Tipo de Atraso\n(dentre os com causa registrada)', fontsize=12)
axes[2].set_ylabel('% dos Voos')
axes[2].tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.show()

# Correlação entre tipos de atraso
print("\nCorrelação entre tipos de atraso:")
print(df_with_cause[delay_cols].rename(columns=dict(zip(delay_cols, delay_labels))).corr().round(2).to_string())
```

# 7. Análise de Aeroportos: Origem e Destino

Aeroportos congestionados tendem a gerar mais atrasos. Vamos identificar os gargalos.

```python
# Aeroportos de ORIGEM com mais atrasos (mínimo 1000 voos)
origin_stats = df_flights.groupby('ORIGIN_AIRPORT').agg(
    total=('ARRIVAL_DELAY', 'count'),
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
origin_stats = origin_stats[origin_stats['total'] >= 1000]
origin_stats['pct_delayed'] *= 100

# Merge com nome do aeroporto
origin_stats = origin_stats.merge(
    df_airports[['IATA_CODE', 'AIRPORT', 'CITY', 'STATE']],
    left_on='ORIGIN_AIRPORT', right_on='IATA_CODE', how='left'
)
origin_stats['label'] = origin_stats['ORIGIN_AIRPORT'].astype(str) + ' - ' + origin_stats['CITY'].fillna('')

# Top 15 aeroportos de DESTINO com maior % de atrasos
dest_stats = df_flights.groupby('DESTINATION_AIRPORT').agg(
    total=('ARRIVAL_DELAY', 'count'),
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
dest_stats = dest_stats[dest_stats['total'] >= 1000]
dest_stats['pct_delayed'] *= 100
dest_stats = dest_stats.merge(
    df_airports[['IATA_CODE', 'AIRPORT', 'CITY']],
    left_on='DESTINATION_AIRPORT', right_on='IATA_CODE', how='left'
)
dest_stats['label'] = dest_stats['DESTINATION_AIRPORT'].astype(str) + ' - ' + dest_stats['CITY'].fillna('')

fig, axes = plt.subplots(1, 2, figsize=(20, 7))

# Top 15 origens com maior % de atrasos
top_origins = origin_stats.nlargest(15, 'pct_delayed').sort_values('pct_delayed')
axes[0].barh(top_origins['label'], top_origins['pct_delayed'], color='#e74c3c', edgecolor='black')
axes[0].set_title('Top 15 Aeroportos de ORIGEM com Maior % de Atrasos', fontsize=12)
axes[0].set_xlabel('% Atrasados')

# Top 15 destinos com maior % de atrasos
top_dests = dest_stats.nlargest(15, 'pct_delayed').sort_values('pct_delayed')
axes[1].barh(top_dests['label'], top_dests['pct_delayed'], color='#3498db', edgecolor='black')
axes[1].set_title('Top 15 Aeroportos de DESTINO com Maior % de Atrasos', fontsize=12)
axes[1].set_xlabel('% Atrasados')

plt.tight_layout()
plt.show()
```

```python
# Top 10 rotas mais populares e suas taxas de atraso
df_flights['ROUTE'] = df_flights['ORIGIN_AIRPORT'].astype(str) + ' → ' + df_flights['DESTINATION_AIRPORT'].astype(str)
route_stats = df_flights.groupby('ROUTE').agg(
    total=('ARRIVAL_DELAY', 'count'),
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
route_stats['pct_delayed'] *= 100

top_routes_vol = route_stats.nlargest(10, 'total').sort_values('total')
top_routes_delay = route_stats[route_stats['total'] >= 500].nlargest(10, 'pct_delayed').sort_values('pct_delayed')

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

axes[0].barh(top_routes_vol['ROUTE'], top_routes_vol['total'] / 1000, color='#3498db', edgecolor='black')
axes[0].set_title('Top 10 Rotas por Volume de Voos', fontsize=12)
axes[0].set_xlabel('Milhares de Voos')

axes[1].barh(top_routes_delay['ROUTE'], top_routes_delay['pct_delayed'], color='#e74c3c', edgecolor='black')
axes[1].set_title('Top 10 Rotas com Maior % de Atrasos (min. 500 voos)', fontsize=12)
axes[1].set_xlabel('% Atrasados')

plt.tight_layout()
plt.show()
```

# 8. Análise de Correlação

Entender a relação entre features numéricas e o `ARRIVAL_DELAY` para guiar a seleção de variáveis no feature engineering.

```python
# Selecionar colunas numéricas relevantes para correlação
numeric_cols = [
    'MONTH', 'DAY', 'DAY_OF_WEEK', 'SCHED_HOUR',
    'SCHEDULED_DEPARTURE', 'DEPARTURE_DELAY', 'TAXI_OUT',
    'SCHEDULED_TIME', 'DISTANCE',
    'TAXI_IN', 'ARRIVAL_DELAY', 'IS_DELAYED'
]
# Filtrar apenas colunas que existem
numeric_cols = [c for c in numeric_cols if c in df_flights.columns]

# Amostra para performance
sample_corr = df_flights[numeric_cols].sample(50000, random_state=42)

corr_matrix = sample_corr.corr()

fig, axes = plt.subplots(1, 2, figsize=(20, 7))

# Heatmap completo
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='RdYlGn',
            center=0, mask=mask, ax=axes[0], linewidths=0.5,
            annot_kws={'size': 9})
axes[0].set_title('Matriz de Correlação (Pearson)\namostra 50k voos', fontsize=12)

# Correlação com ARRIVAL_DELAY ordenada
corr_target = corr_matrix['ARRIVAL_DELAY'].drop('ARRIVAL_DELAY').sort_values()
colors_corr = ['#e74c3c' if x > 0 else '#2ecc71' for x in corr_target]
axes[1].barh(corr_target.index, corr_target.values, color=colors_corr, edgecolor='black')
axes[1].axvline(x=0, color='black', linewidth=1)
axes[1].set_title('Correlação de Cada Feature com ARRIVAL_DELAY', fontsize=12)
axes[1].set_xlabel('Correlação de Pearson')

plt.tight_layout()
plt.show()

print("\nObservação: DEPARTURE_DELAY tem correlação alta com ARRIVAL_DELAY,")
print("mas é informação de runtime (não disponível no momento da previsão).")
print("No modelo preditivo (pré-voo), ela NÃO pode ser usada como feature.")
```

# 9. Análise de Outliers e Relação entre Atraso na Partida e Chegada

O `DEPARTURE_DELAY` tem correlação fortíssima com o `ARRIVAL_DELAY`. Entender essa relação e os outliers é crítico para o tratamento no feature engineering.

# 10. Análise de Distância e Tempo de Voo

A distância e o tempo programado de voo podem influenciar a probabilidade de atraso.

```python
# Criar faixas de distância
distance_bins = [0, 500, 1000, 1500, 2000, 3000, 10000]
distance_labels = ['< 500', '500-1000', '1000-1500', '1500-2000', '2000-3000', '> 3000']

df_flights['DISTANCE_BIN'] = pd.cut(df_flights['DISTANCE'], bins=distance_bins, labels=distance_labels)

by_distance = df_flights.groupby('DISTANCE_BIN', observed=True).agg(
    total=('ARRIVAL_DELAY', 'count'),
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
by_distance['pct_delayed'] *= 100

# Faixas de tempo programado
time_bins = [0, 60, 120, 180, 240, 360, 1000]
time_labels = ['< 1h', '1h-2h', '2h-3h', '3h-4h', '4h-6h', '> 6h']
df_flights['SCHED_TIME_BIN'] = pd.cut(df_flights['SCHEDULED_TIME'], bins=time_bins, labels=time_labels)

by_time = df_flights.groupby('SCHED_TIME_BIN', observed=True).agg(
    total=('ARRIVAL_DELAY', 'count'),
    avg_delay=('ARRIVAL_DELAY', 'mean'),
    pct_delayed=('IS_DELAYED', 'mean')
).reset_index()
by_time['pct_delayed'] *= 100

fig, axes = plt.subplots(2, 2, figsize=(18, 10))

# % atrasados por distância
axes[0, 0].bar(by_distance['DISTANCE_BIN'].astype(str), by_distance['pct_delayed'],
               color='#3498db', edgecolor='black')
axes[0, 0].set_title('% de Atrasos por Faixa de Distância (milhas)', fontsize=12)
axes[0, 0].set_xlabel('Distância (milhas)')
axes[0, 0].set_ylabel('% Atrasados')

# Atraso médio por distância
axes[0, 1].bar(by_distance['DISTANCE_BIN'].astype(str), by_distance['avg_delay'],
               color='#9b59b6', edgecolor='black')
axes[0, 1].set_title('Atraso Médio (min) por Faixa de Distância', fontsize=12)
axes[0, 1].set_xlabel('Distância (milhas)')
axes[0, 1].set_ylabel('Minutos de Atraso')

# % atrasados por duração programada
axes[1, 0].bar(by_time['SCHED_TIME_BIN'].astype(str), by_time['pct_delayed'],
               color='#e67e22', edgecolor='black')
axes[1, 0].set_title('% de Atrasos por Duração Programada do Voo', fontsize=12)
axes[1, 0].set_xlabel('Duração Programada')
axes[1, 0].set_ylabel('% Atrasados')

# Volume por faixa de distância
axes[1, 1].bar(by_distance['DISTANCE_BIN'].astype(str), by_distance['total'] / 1000,
               color='#95a5a6', edgecolor='black')
axes[1, 1].set_title('Volume de Voos por Faixa de Distância', fontsize=12)
axes[1, 1].set_xlabel('Distância (milhas)')
axes[1, 1].set_ylabel('Milhares de Voos')

plt.tight_layout()
plt.show()

print("\nEstatísticas de DISTANCE:")
print(df_flights['DISTANCE'].describe().round(1).to_string())
```

# 11. Heatmap: Mês × Dia da Semana

Visualizar padrões combinados de atraso ajuda a identificar interações importantes para o feature engineering.

```python
# Pivot: Mês x Dia da Semana → % de atrasos
pivot_delay = df_flights.pivot_table(
    values='IS_DELAYED', index='MONTH', columns='DAY_OF_WEEK', aggfunc='mean'
) * 100

# Renomear índices
pivot_delay.index = [month_names[m] for m in pivot_delay.index]
pivot_delay.columns = [day_names[d] for d in pivot_delay.columns]

# Pivot: Hora x Dia da Semana → atraso médio
pivot_hour_dow = df_flights.pivot_table(
    values='ARRIVAL_DELAY', index='SCHED_HOUR', columns='DAY_OF_WEEK', aggfunc='mean'
)
pivot_hour_dow.columns = [day_names[d] for d in pivot_hour_dow.columns]

fig, axes = plt.subplots(1, 2, figsize=(20, 6))

# Heatmap mês × dia da semana
sns.heatmap(pivot_delay, annot=True, fmt='.1f', cmap='RdYlGn_r',
            ax=axes[0], linewidths=0.5, cbar_kws={'label': '% Atrasados'})
axes[0].set_title('% de Voos Atrasados: Mês × Dia da Semana', fontsize=13)
axes[0].set_xlabel('Dia da Semana')
axes[0].set_ylabel('Mês')

# Heatmap hora × dia da semana
sns.heatmap(pivot_hour_dow, cmap='RdYlGn_r', ax=axes[1],
            linewidths=0.3, cbar_kws={'label': 'Atraso Médio (min)'})
axes[1].set_title('Atraso Médio (min): Hora do Dia × Dia da Semana', fontsize=13)
axes[1].set_xlabel('Dia da Semana')
axes[1].set_ylabel('Hora do Dia (Partida)')

plt.tight_layout()
plt.show()
```
